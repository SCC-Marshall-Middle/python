# Comments
- Used to clarify code and are ignored by Python
- Start with the hash character ‘#’ and extend to the end of the physical line
- Allow you to write in English within your program

````python
# This line is a comment.
print("This line is not a comment, it is code.")
````

## Multi-line Comments
- Allows a comment to extend past 1 line
- Created using 2 sets of 3 quotation marks (""")

````python
"""
This is a multi-line comment.
This comment can extend past one line.
"""
# This line is a comment.
# With multi-line comments, you don’t have to type # multiple times
print("This line is not a comment, it is code.")
````

## What makes a good comment?
- It is short and to the point, but a complete thought. 
- It explains your thinking, so that:
    - When you return to the code later you will understand how you were approaching the problem.
    - Others who work with your code will understand your overall approach to a problem.
    - It explains particularly difficult sections of code in detail.
- Don’t write comments because you feel like you have to--make them helpful.

## When should you write comments?
- When you have to think about code before writing it.
- When you are likely to forget later exactly how you were approaching a problem.
- When there is more than one way to solve a problem.
- When others are unlikely to anticipate your way of thinking about a problem.

# Numbers
- 2 numeric types in Python
    - int
    - float
- Variables of numeric types are created when you assign numbers to them

```python
1		# int
2.8		# float
```

#### Int (integer)
- A whole number, positive or negative, without decimals, of unlimited length
- Ex: 1, 35467, -94830

#### Float (floating-point)
- A number, positive or negative, containing  one or more decimals
- Can also be scientific numbers with an ‘e’ to indicate powers of 10
- Ex: 1.0, -35.59, 35e3
Here, 35e3 = 35*10^3 = 35*1000 = 35000

> You cannot put commas in your numbers. 3,400 will not work!

## Casting
- Done when you need to change the data type of a variable
- Constructor functions:
    - int()- rounds down a float to an int (ex 2.8 → 2), or converts a string to an int (ex "32"→32)
    - float()- adds a decimal point to an int (ex 2→2.0), or converts a string to a float                      (ex "1.6"→1.6)
    - str()- converts ints/floats to strings (ex 24→"24", 2.4→"2.4")
- Remember that a number with quotes is a string, and a number without quotes can be an int or a float
    - A number with quotes and a number without quotes are not the same thing!
> Why do we need casting (hint: do 2 + 4 and "2" + "4" give the same thing)?

### Examples

````python
# integer casting
int(1)	# converted to 1
int(2.8)	# converted to 2
int("3")	# converted to 3

# float casting
float(1)		# converted to 1.0
float(2.8)	# converted to 2.8
float("3")	# converted to 3.0
float("4.2")	# converted to 4.2

# string casting
str("s1")	# converted to ‘s1’
str(2)	# converted to ‘2’
str(3.0)	# converted to ‘3.0’
````
